```python
def calculate_returns():
    try:
        # 1. Accept the number of time steps/episodes
        num_steps = int(input("Enter the number of time steps: "))

        rewards = []
        # 2. Accept the reward/return value for each time step
        for i in range(num_steps):
            reward = float(input(f"Enter the reward for time step {i}: "))
            rewards.append(reward)

        # 3. Accept the discount factor
        gamma = float(input("Enter the discount factor (gamma): "))

        # 4. Calculate the undiscounted return (sum of all rewards)
        undiscounted_return = sum(rewards)

        # 5. Calculate the discounted return (sum of gamma^t * R_{t+1})
        discounted_return = 0
        for t, r in enumerate(rewards):
            discounted_return += (gamma ** t) * r

        # 6. Display results
        print("\n--- Results ---")
        print(f"Entered reward values: {rewards}")
        print(f"Undiscounted return: {undiscounted_return}")
        print(f"Discounted return: {discounted_return}")

    except ValueError:
        print("Invalid input. Please enter numerical values.")

if __name__ == "__main__":
    calculate_returns()
```

```python
# grid_env_det.py

# a. Define State Space
# States are represented by their coordinates (row, col)
# S0=(0,0), S1=(0,1), G=(0,2), S3=(1,1), S4=(1,2)
STATES = ['S0', 'S1', 'G', 'S3', 'S4']

STATE_COORDS = {
    'S0': (0, 0),
    'S1': (0, 1),
    'G':  (0, 2),
    'X':  (1, 0), # Obstacle
    'S3': (1, 1),
    'S4': (1, 2)
}

# Reverse mapping to easily find the state name from coordinates
COORD_TO_STATE = {v: k for k, v in STATE_COORDS.items()}

# b. Define Action Space
ACTIONS = ['Up', 'Down', 'Left', 'Right']

# c. Define Reward Function
def get_reward(state):
    """Returns the reward for entering a given state."""
    if state == 'G':
        return 10
    elif state == 'X':
        return -5
    else:
        return -1 # Normal state

# d. Define Discount Factor
DISCOUNT_FACTOR = 0.9

# e. & f. Define Transition Function and Estimate Next State
def transition(current_state, action):
    """
    Returns the next state given a current state and an action.
    Transitions are deterministic.
    """
    # Terminal states
    if current_state in ['G', 'X']:
        return current_state

    r, c = STATE_COORDS[current_state]

    # Determine intended next coordinates
    if action == 'Up':
        next_r, next_c = r - 1, c
    elif action == 'Down':
        next_r, next_c = r + 1, c
    elif action == 'Left':
        next_r, next_c = r, c - 1
    elif action == 'Right':
        next_r, next_c = r, c + 1
    else:
        return current_state

    # Check boundaries (Grid is 2 rows by 3 columns)
    if next_r < 0 or next_r > 1 or next_c < 0 or next_c > 2:
        # If action takes agent outside grid, stay in same state
        next_r, next_c = r, c

    next_coord = (next_r, next_c)
    next_state_name = COORD_TO_STATE.get(next_coord, current_state)

    # Obstacle check: Based on the transition table, agent cannot move into X
    # (e.g., S0 Down -> S0, S3 Left -> S3)
    if next_state_name == 'X':
        return current_state

    return next_state_name

# --- Testing the Environment ---
if __name__ == "__main__":
    print(f"State Space (S): {STATES}")
    print(f"Action Space (A): {ACTIONS}")
    print(f"Discount Factor (gamma): {DISCOUNT_FACTOR}\n")

    print("--- Transition Tests ---")
    test_cases = [
        ('S0', 'Right'),
        ('S1', 'Right'),
        ('S0', 'Down'),  # Should hit obstacle and stay in S0
        ('S4', 'Up'),    # Should reach Goal
        ('S3', 'Left')   # Should hit obstacle and stay in S3
    ]

    for s, a in test_cases:
        next_s = transition(s, a)
        reward = get_reward(next_s)
        print(f"Current State: {s} | Action: {a:5} -> Next State: {next_s} | Reward: {reward}")
```
